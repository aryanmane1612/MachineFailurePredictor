```python
from pathlib import Path
import mat73

ROOT = (
    Path.home()
    / "Desktop"
    / "MachineFailurePredictor"
    / "data"
    / "raw"
    / "paderborn"
)

files = sorted(ROOT.rglob("*.mat"))

print("Dataset folder:", ROOT)
print("Total MAT files:", len(files))

if not files:
    print("No MAT files found. Check the dataset path.")
else:
    for path in files[:3]:
        print("\nFILE:", path.name)

        try:
            data = mat73.loadmat(str(path))
            print("Top-level keys:", list(data.keys()))

            def inspect(obj, prefix=""):
                if isinstance(obj, dict):
                    for key, value in obj.items():
                        inspect(value, prefix + "/" + str(key))
                elif hasattr(obj, "shape"):
                    print(
                        prefix,
                        "| shape:", obj.shape,
                        "| dtype:", getattr(obj, "dtype", type(obj))
                    )

            inspect(data)

        except Exception as error:
            print("ERROR:", repr(error))
```

```python
from pathlib import Path
from scipy.io import loadmat

ROOT = (
    Path.home()
    / "Desktop"
    / "MachineFailurePredictor"
    / "data"
    / "raw"
    / "paderborn"
)

files = sorted(ROOT.rglob("*.mat"))

print("Total MAT files:", len(files))

for path in files[:3]:
    print("\nFILE:", path.name)

    try:
        data = loadmat(str(path))

        keys = [
            key for key in data.keys()
            if not key.startswith("__")
        ]

        print("Top-level keys:", keys)

        for key in keys:
            value = data[key]
            print(
                "Key:", key,
                "| type:", type(value),
                "| shape:", getattr(value, "shape", None),
                "| dtype:", getattr(value, "dtype", None)
            )

    except Exception as error:
        print("ERROR:", repr(error))
```

```python
from pathlib import Path
from scipy.io import loadmat
import numpy as np

ROOT = (
    Path.home()
    / "Desktop"
    / "MachineFailurePredictor"
    / "data"
    / "raw"
    / "paderborn"
)

file_path = next(ROOT.rglob("N09_M07_F10_K001_1.mat"))
data = loadmat(str(file_path))

# Get the MATLAB structure
mat_name = file_path.stem
record = data[mat_name][0, 0]

print("File:", file_path.name)
print("Fields:", record.dtype.names)

# Inspect each field
for field in record.dtype.names:
    value = record[field]

    print(f"\n--- {field} ---")
    print("Type:", type(value))
    print("Shape:", getattr(value, "shape", None))
    print("Dtype:", getattr(value, "dtype", None))

    if isinstance(value, np.ndarray):
        print("First few values:")
        print(value.reshape(-1)[:5])
```

```python
import json
import os
import joblib
import pandas as pd

project_path = r"C:\Users\aryan\Desktop\MachineFailurePredictor"

# Load the saved feature column names
columns_path = os.path.join(project_path, "models", "feature_columns.json")

with open(columns_path, "r") as f:
    model_features = json.load(f)

# Compare dataset columns with model columns
dataset_features = [
    col for col in df.columns
    if col not in ["file", "bearing_id", "label", "condition"]
]

print("Dataset feature count:", len(dataset_features))
print("Model feature count:", len(model_features))

print("\nFeatures missing from dataset:")
print(sorted(set(model_features) - set(df.columns)))

print("\nFeatures missing from model:")
print(sorted(set(dataset_features) - set(model_features)))
```

```python
import joblib
import json
import os
import pandas as pd

project_path = r"C:\Users\aryan\Desktop\MachineFailurePredictor"

# Load the saved model and expected feature names
model = joblib.load(
    os.path.join(project_path, "models", "machine_failure_svm.joblib")
)

with open(
    os.path.join(project_path, "models", "feature_columns.json"), "r"
) as f:
    model_features = json.load(f)

# Prepare the feature data in the exact expected order
X = df[model_features]

# Predict a few existing recordings
sample = X.head(10)
predictions = model.predict(sample)

results = pd.DataFrame({
    "file": df.loc[sample.index, "file"],
    "actual_condition": df.loc[sample.index, "condition"],
    "predicted_label": predictions,
    "predicted_condition": [
        "Damaged" if p == 1 else "Healthy" for p in predictions
    ]
})

display(results)
```

```python
# Select 5 healthy and 5 damaged recordings
sample_healthy = df[df["label"] == 0].head(5)
sample_damaged = df[df["label"] == 1].head(5)

test_sample = pd.concat([sample_healthy, sample_damaged])

# Predict using the saved model
X_test_sample = test_sample[model_features]
predictions = model.predict(X_test_sample)

results = pd.DataFrame({
    "file": test_sample["file"].values,
    "actual_condition": test_sample["condition"].values,
    "predicted_condition": [
        "Damaged" if p == 1 else "Healthy"
        for p in predictions
    ]
})

display(results)
```
